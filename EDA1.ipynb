```python
import pandas as pd
import numpy as np
```

```python
dataFrame1=pd.DataFrame({"StudentID":[1,3,5,7,9],"Score":[89,39,50,97,22]})
dataFrame2=pd.DataFrame({"StudentID":[2,4,6,8,10],"Score":[98,93,44,77,69]})
vertical_concat=pd.concat([dataFrame1,dataFrame2],ignore_index=True)
print("Vertical Concatenation:\n",vertical_concat)
```

```python
horizontal_concat=pd.concat([dataFrame1,dataFrame2],axis=1)
print("Horizontal Concatenation:\n",horizontal_concat)
```

```python
dfSE=pd.DataFrame({"StudentID":[2,4,6,8,10,12],"ScoreSE":[98,93,44,77,69,56]})
dfML=pd.DataFrame({"StudentID":[1,2,3,4,5,6,8,10],"ScoreML":[39,93,49,44,55,66,77,88]})
inner_join=pd.merge(dfSE,dfML,on="StudentID",how="inner")
print("\nInner join:\n",inner_join)
```

```python
left_join=pd.merge(dfSE,dfML,on="StudentID",how="left")
print("\\nLeft join:\n",left_join)
```

```python
right_join=pd.merge(dfSE,dfML,on="StudentID",how="right")
print("\nRight join:\n",right_join)
```

```python
outer_join=pd.merge(dfSE,dfML,on="StudentID",how="inner")
print("\nOuter join:\n",outer_join)
```

```python
sales=pd.DataFrame({"Quantity":[100,250,300,150],"UnitPrice":[50,60,45,80]})
sales["TotalPrice"]=sales["Quantity"]*sales["UnitPrice"]
print("\nSales Data:\n",sales)
```

```python
print("\nDescriptive Statistics:")
print(sales.describe())
```

```python
company_df=pd.DataFrame({"Company":["ABC","XYZ","ABC","PQR","XYZ","ABC"]})
print("\nCompany Counts:\n",company_df["Company"].value_counts())
```

```python
data=np.arange(15).reshape((3,5))
dframe=pd.DataFrame(data,index=["Rainfall","Humanity","Wind"],columns=["Bergen","Oslo","Trodheim","Stavanger","Kristiansand"])
print("\nOriginal DataFrame:\n",dframe)

stacked=dframe.stack()
print("\nStacked Data:\n",stacked)
```

```python
unstacked=stacked.unstack()
print("Unstacked data:\n",unstacked)
```

```python
series1=pd.Series([0, 111, 222, 333],index=['zeros', 'ones', 'twos', 'threes'])
series2=pd.Series([444, 555, 666],index=['fours', 'fives', 'sixs'])
frame=pd.concat([series1, series2],keys=['Number1', 'Number2'])
print("\nHierarchical Series:\n",frame)
print("\nUnstacked Series:\n",frame.unstack())
```
