```python
import pandas as pd
import numpy as np
```

```python
salaries=[10,30,25,60,49,97,80,40,24,13,50,35]
bins=[0,25,50,75,100]
intervals=pd.cut(salaries,bins)
intervals
```

```python
pd.value_counts(intervals)
```

```python
bin_names=["Least","Decent","Good","Rich"]
intervals=pd.cut(salaries,bins,labels=bin_names)
intervals
```

```python
intervals=pd.cut(salaries,5)
intervals
```

```python
pd.cut(np.random.rand(25),5,precision=2)
```

```python
randomNum=np.random.rand(1000)
category=pd.qcut(randomNum,4)
category
```

```python
pd.value_counts(category)
```

```python
from sklearn.preprocessing import MinMaxScaler,StandardScaler,RobustScaler
```

```python
data={
 "Age":[10, 12, 11, 13, 12],
 "Height":[97, 95, 110, 120, 200]
}
df=pd.DataFrame(data)
df
```

```python
min_max_scaler=MinMaxScaler()
df_min_max=pd.DataFrame(min_max_scaler.fit_transform(df),columns=df.columns)
print("Min-Max Scaled Data:\n",df_min_max,"\n")
```

```python
standard_scaler=StandardScaler()
df_standard=pd.DataFrame(standard_scaler.fit_transform(df),columns=df.columns)
print("Standardized Data:\n",df_standard,"\n")
```
