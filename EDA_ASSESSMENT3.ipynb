```python
import pandas as pd
import numpy as np
```

```python
df1=pd.DataFrame({'Restaurant':['Mcd','KFC','Burger King','SFC','Subway','Jacks','Bubee'],'Sales in millions':[6,5,0.1,43,15,6.7,0.5]})
df2=pd.DataFrame({'Restaurant':['Hardees','Lakshmi caterers','Darling','Bubee','Jimmy John','Hatam','Albaik'],'Sales in millions':[14,22,9,0.5,0.78,22,1]})
```

```python
df=pd.concat([df1,df2],axis=1,ignore_index=True)
df
```

```python
dfMerge=df1.merge(df2,how='inner')
dfMerge
```

```python
dfMerge=df1.merge(df2,how='left')
dfMerge
```

```python
dfMerge=df1.merge(df2,how='right')
dfMerge
```
